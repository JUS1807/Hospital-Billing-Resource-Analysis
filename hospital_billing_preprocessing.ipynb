```python
# Variante 2 löschen und CSV speichern
variant_2_cases = df[df["Variant index"] == 2]["Case ID"].unique()
df_clean = df[~df["Case ID"].isin(variant_2_cases)].copy()
print("Original Cases:", df["Case ID"].nunique())
print("Cleaned Cases:", df_clean["Case ID"].nunique())
print("Removed Cases:", len(variant_2_cases))
```

```python
df_clean.to_csv("Hospital Billing - Event Log_cleaned.csv", index=False)
```
